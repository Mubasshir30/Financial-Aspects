```python
import numpy as np

def call_option_price(S, K, r, sigma, T, simulations):
    """
    Calculates the price of a European call option using Monte Carlo simulation.

    Parameters:
    S (float): Current price of the underlying asset
    K (float): Strike price of the option
    r (float): Risk-free interest rate (annual)
    sigma (float): Volatility of the underlying asset (annual)
    T (float): Time to expiration (in years)
    simulations (int): Number of Monte Carlo simulations to run

    Returns:
    float: Estimated price of the call option
    """
    dt = 1 / 365.0  # Time step: one day in years
    n = int(T / dt)  # Number of time steps

    # Simulate price paths (vectorized for efficiency)
    S_t = S * np.exp(np.cumsum(
        (r - 0.5 * sigma**2) * dt + sigma * np.sqrt(dt) * np.random.normal(0, 1, (n, simulations)),
        axis=0
    ))

    # Payoff at expiration for each simulation
    payoff = np.maximum(S_t[-1] - K, 0)

    # Discount payoffs to present value and average
    call_price = np.exp(-r * T) * np.mean(payoff)
    return call_price

# Example usage
S = 110  # Current stock price
K = 100  # Strike price
r = 0.01  # Risk-free rate (1%)
sigma = 0.2  # Volatility (20%)
T = 30/365  # Time to expiration (30 days in years)
simulations = 100000  # Number of simulations

price = call_option_price(S, K, r, sigma, T, simulations)
print(f"Estimated Call Option Price: {price:.4f}")
```
