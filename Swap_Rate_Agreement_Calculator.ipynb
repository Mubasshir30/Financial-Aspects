```python
import math

def discount_factor(rate, t):
    """Simple compounding discount factor."""
    return 1 / (1 + rate * t)

def swap_pricing_interactive():
    print("=== Swap Rate Agreement Calculator ===")
    notional = float(input("Enter Notional Amount (e.g., 5000000): "))
    pay_freq = int(input("Enter Payment Frequency per Year (e.g., 4 for quarterly, 2 for semiannual): "))

    accrual = 1 / pay_freq
    n_periods = int(input("Enter number of periods (e.g., 4 for 1 year quarterly): "))

    spot_rates = {}
    for i in range(1, n_periods + 1):
        t = i * accrual
        r = float(input(f"Enter annualized spot rate for {int(t*360)} days (as decimal, e.g., 0.035 for 3.5%): "))
        spot_rates[t] = r

    # Step 1: Discount factors
    dfs = {t: discount_factor(spot_rates[t], t) for t in spot_rates}

    # Step 2: Fixed rate
    numerator = 1 - dfs[max(spot_rates.keys())]
    denominator = sum(dfs[t] * accrual for t in spot_rates)
    fixed_rate = numerator / denominator

    # Step 3: Fixed payment
    fixed_payment = notional * fixed_rate * accrual

    # Step 4: First floating payment
    first_t = min(spot_rates.keys())
    first_float_payment = notional * spot_rates[first_t] * accrual

    # Step 5: Net settlement
    first_net = first_float_payment - fixed_payment

    print("\n=== Results ===")
    print(f"Fixed Rate (%): {round(fixed_rate * 100, 4)}")
    print(f"Fixed Payment: {round(fixed_payment, 2)}")
    print(f"First Floating Payment: {round(first_float_payment, 2)}")
    print(f"First Net Payment: {round(first_net, 2)}")

# ---------------- Run ----------------
if __name__ == "__main__":
    swap_pricing_interactive()
```
